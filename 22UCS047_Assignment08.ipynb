```python
from sklearn.datasets import make_blobs, make_moons, make_circles
import pandas as pd
import matplotlib.pyplot as plt
```

## Data Generation

```python
# X1, y1 = make_blobs(n_samples=200, centers=5, random_state=20)
X1, y1 = make_blobs(n_samples=300, centers=4, random_state=0, cluster_std=0.6)
X2, y2 = make_moons(n_samples=200, noise=0.1, random_state=13)
X3, y3 = make_circles(n_samples=200, noise=0.1, random_state=42, factor=0.2)

df1 = pd.DataFrame(X1, columns=['x', 'y'])
df2 = pd.DataFrame(X2, columns=['x', 'y'])
df3 = pd.DataFrame(X3, columns=['x', 'y'])
```

```python
plt.scatter(df1['x'], df1['y'], c=y1)
plt.show()
plt.scatter(df2['x'], df2['y'], c=y2)
plt.show()
plt.scatter(df3['x'], df3['y'], c=y3)
plt.show()
```

## Hierarchical Clustering

```python
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering
```

### Linkage: Single

```python
linkage_data = linkage(df1, method='single')
dendrogram(linkage_data)
# plt.rcParams["figure.figsize"] = (9, 6)
plt.show()
```

```python
hc= AgglomerativeClustering(n_clusters=4, linkage='single')
y1_pred_single= hc.fit_predict(df1)

plt.scatter(df1['x'], df1['y'], c=y1_pred_single)
plt.show()
```

```python
linkage_data = linkage(df2, method='single')
dendrogram(linkage_data)
plt.show()
```

```python
hc= AgglomerativeClustering(n_clusters=2, linkage='single')
y2_pred_single= hc.fit_predict(df2)

plt.scatter(df2['x'], df2['y'], c=y2_pred_single)
plt.show()
```

```python
linkage_data = linkage(df3, method='single')
dendrogram(linkage_data)
plt.show()
```

```python
hc= AgglomerativeClustering(n_clusters=2, linkage='single')
y3_pred_single= hc.fit_predict(df3)

plt.scatter(df3['x'], df3['y'], c=y3_pred_single)
plt.show()
```

## Density-Based Clustering

```python
from sklearn.cluster import DBSCAN
```

```python
clustering = DBSCAN(eps=0.6, min_samples=5).fit(df1)
y1_pred_dbscan = clustering.labels_
plt.scatter(df1['x'], df1['y'], c=clustering.labels_)
plt.show()
```

```python
clustering = DBSCAN(eps=0.2, min_samples=7).fit(df2)
y2_pred_dbscan = clustering.labels_
plt.scatter(df2['x'], df2['y'], c=clustering.labels_)
plt.show()
```

```python
clustering = DBSCAN(eps=0.3, min_samples=3).fit(df3)
y3_pred_dbscan = clustering.labels_
plt.scatter(df3['x'], df3['y'], c=clustering.labels_)
plt.show()
```

## Evaluation

```python
from sklearn.metrics import silhouette_score
```

```python
print(silhouette_score(df1, y1_pred_single))
print(silhouette_score(df2, y2_pred_single))
print(silhouette_score(df3, y3_pred_single))
print(silhouette_score(df1, y1_pred_complete))
print(silhouette_score(df2, y2_pred_complete))
print(silhouette_score(df3, y3_pred_complete))
print(silhouette_score(df2, y2_pred_average))
print(silhouette_score(df3, y3_pred_average))
print(silhouette_score(df1, y1_pred_average))
```

```python
print(silhouette_score(df1, y1_pred_dbscan))
print(silhouette_score(df2, y2_pred_dbscan))
print(silhouette_score(df3, y3_pred_dbscan))
```
