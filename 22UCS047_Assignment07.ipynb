```python
from sklearn.datasets import make_blobs, make_moons, make_circles
import pandas as pd
import matplotlib.pyplot as plt
```

## Data Generation

```python
X1, y1 = make_blobs(n_samples=200, centers=5, random_state=13)
X2, y2 = make_moons(n_samples=200, noise=0.1, random_state=13)
X3, y3 = make_circles(n_samples=200, noise=0.1, random_state=42, factor=0.2)

df1 = pd.DataFrame(X1, columns=['x', 'y'])
df2 = pd.DataFrame(X2, columns=['x', 'y'])
df3 = pd.DataFrame(X3, columns=['x', 'y'])
```

```python
plt.scatter(df1['x'], df1['y'], c=y1)
plt.show()
plt.scatter(df2['x'], df2['y'], c=y2)
plt.show()
plt.scatter(df3['x'], df3['y'], c=y3)
plt.show()
```

## Model

```python
from sklearn.cluster import KMeans

model1 = KMeans(n_clusters=5)
model1.fit(df1[['x', 'y']])
X1_test = X1
y1_test = model1.predict(X1_test)
plt.scatter(df1['x'], df1['y'], c=y1_test)
plt.show()

model2 = KMeans(n_clusters=2)
model2.fit(df2[['x', 'y']])
X2_test = X2
y2_test = model2.predict(X2_test)
plt.scatter(df2['x'], df2['y'], c=y2_test)
plt.show()

model3 = KMeans(n_clusters=2)
model3.fit(df3[['x', 'y']])
X3_test = X3
y3_test = model3.predict(X3_test)
plt.scatter(df3['x'], df3['y'], c=y3_test)
plt.show()
```

## Evaluation

```python
from sklearn.metrics import silhouette_score

print(silhouette_score(df1, y1_test))
print(silhouette_score(df2, y2_test))
print(silhouette_score(df3, y3_test))
```

```python
print(model1.inertia_)
print(model2.inertia_)
print(model3.inertia_)
```

```python
inert = [0 for i in range(1, 51)]
for i in range(50):
    model = KMeans(n_clusters=i+1)
    model.fit(df1[['x', 'y']])
    inert[i] = model.inertia_

plt.plot(range(1, 51), inert)
plt.xticks(range(1, 51, 2))
plt.show()
```
